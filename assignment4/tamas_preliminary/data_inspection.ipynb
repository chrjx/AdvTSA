```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))
plt.scatter(data["toy"], data["p"], s=2, alpha=0.3)
plt.xlabel("Time of Year (days)")
plt.ylabel("Power (kW)")
plt.title("Power production vs Time of Year")
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(data["Ws1"], data["p"], s=2, alpha=0.3)
plt.xlabel("Wind Speed (m/s)")
plt.ylabel("Power (kW)")
plt.title("Power curve (raw)")
plt.show()
```

```python
data["Ws_bin"] = pd.cut(data["Ws1"], bins=np.arange(0, 36, 1))

power_curve = data.groupby("Ws_bin")["p"].mean()

power_curve.plot(figsize=(7,4))
plt.xlabel("Wind speed bin (m/s)")
plt.ylabel("Mean power (kW)")
plt.title("Estimated power curve")
plt.show()
```

```python
data["Wd1_rad"] = np.deg2rad(data["Wd1"])
data["Wd1_sin"] = np.sin(data["Wd1_rad"])
data["Wd1_cos"] = np.cos(data["Wd1_rad"])
```

```python
data ["Wd_sector"] = pd.cut(data["Wd1"], bins=np.arange(0, 361, 30))

dir_power = data.groupby("Wd_sector")["p"].mean()

dir_power.plot(kind="bar", figsize=(10,4))
plt.ylabel("Mean power (kW)")
plt.title("Mean power vs wind direction sector")
plt.show()
data.drop(columns=["Wd_sector"], inplace=True)
```

```python
for lag in [1, 2, 3]:
    data[f"p_lag{lag}"] = data["p"].shift(lag)
```

```python
from pandas.plotting import autocorrelation_plot

autocorrelation_plot(data["p"].dropna())
plt.show()
```

```python
data["time"] = data.index
data["hour"] = pd.to_datetime(data["time"]).dt.hour

data.groupby("hour")["p"].mean().plot()
plt.xlabel("Hour of day")
plt.ylabel("Mean power")
plt.title("Diurnal variation in power")
plt.show()
```
